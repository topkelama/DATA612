## Research Discussion Assignment 2

DATA 612 - Recommender Systems

 Chhiring Lama

One of the most interesting parts of this talk for me was learning how Spotify uses collaborative filtering to recommend music. I liked the simple idea that if two users have similar listening habits, they may enjoy songs that the other person has discovered. It was also interesting to see how Spotify uses listening history as implicit feedback instead of relying on users to provide ratings.
Another takeaway was the use of matrix factorization and ALS to uncover hidden relationships between users and songs. While some of the mathematics was advanced, it was helpful to see how these techniques are applied in a real-world recommendation system.
The second half of the talk highlighted the challenges of working with data at a massive scale. I found the comparison between Hadoop and Spark particularly interesting because Spark's in-memory processing significantly reduced the time needed for iterative algorithms like ALS. Overall, the talk showed that building recommendation systems requires both strong machine learning techniques and efficient data processing systems.
